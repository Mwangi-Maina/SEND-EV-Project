# 11A - Get ACN Data

Saves 2019 ACN sessions as site CSVs.

Modes:

- `"api"`: download from ACN.
- `"manual"`: read local files.

The end date is exclusive, so 31 December 2019 is included.

## Files

```text
Data/
└── ACN/
    ├── acquired/
    │   ├── caltech_2019_raw.jsonl
    │   ├── caltech_2019_sessions.csv
    │   ├── jpl_2019_raw.jsonl
    │   ├── jpl_2019_sessions.csv
    │   ├── office001_2019_raw.jsonl
    │   ├── office001_2019_sessions.csv
    │   ├── acquisition_manifest.csv
    │   ├── acquisition_chunk_audit.csv
    │   └── acquisition_request_audit.csv
    ├── chunks/
    │   ├── caltech/
    │   ├── jpl/
    │   └── office001/
    └── manual/
        └── optional manually downloaded source files
```

11B reads files ending in
`_2019_sessions.csv`.

````python
from datetime import datetime, timedelta, timezone
from email.utils import format_datetime, parsedate_to_datetime
from getpass import getpass
from pathlib import Path
from typing import Any, Iterator
from urllib.parse import urljoin
import ast as python_ast
import hashlib
import json
import os
import platform
import random
import re
import sys
import time
import warnings

import numpy as np
import pandas as pd
import requests

warnings.filterwarnings("ignore", category=FutureWarning)

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 180)
pd.set_option("display.max_colwidth", 140)


def find_project_root(start: Path | None = None) -> Path:
    current = Path(start or Path.cwd()).resolve()

    for candidate in [current, *current.parents]:
        if (candidate / "Data").exists() and (candidate / "Notebooks").exists():
            return candidate

    raise FileNotFoundError(
        "Could not locate the SEND-EV project root. "
        "Open this notebook from a repository containing Data/ and Notebooks/."
    )


ROOT = find_project_root()
DATA_DIR = ROOT / "Data"
ACN_DIR = DATA_DIR / "ACN"
ACQUIRED_DIR = ACN_DIR / "acquired"
CHUNK_DIR = ACN_DIR / "chunks"
MANUAL_DIR = ACN_DIR / "manual"

for directory in [ACN_DIR, ACQUIRED_DIR, CHUNK_DIR, MANUAL_DIR]:
    directory.mkdir(parents=True, exist_ok=True)

print("Project root:", ROOT)
print("Python:", sys.executable)
print("Python version:", platform.python_version())
print("Acquired-data directory:", ACQUIRED_DIR)
````

## 2. Config

````python
SITES = ("caltech", "jpl", "office001")

# Source.
#   "api" downloads data.
#   "manual" reads local files.
ACQUISITION_MODE = "api"

EXTRACTION_START = datetime(2019, 1, 1, tzinfo=timezone.utc)
EXTRACTION_END = datetime(2020, 1, 1, tzinfo=timezone.utc)

API_BASE_URL = "https://ev.caltech.edu/api/v1/"
MIN_ENERGY_KWH = 0.0
SORT_ARGUMENT = "-connectionTime"

# Small chunks retry better.
SITE_CHUNK_DAYS = {
    "caltech": 7,
    "jpl": 7,
    "office001": 31,
}

CONNECT_TIMEOUT_SECONDS = 20
READ_TIMEOUT_SECONDS = 90
MAX_REQUEST_ATTEMPTS = 6
BASE_RETRY_SECONDS = 3.0
MAX_RETRY_SECONDS = 90.0
PAUSE_BETWEEN_PAGES_SECONDS = 0.25
PAUSE_BETWEEN_CHUNKS_SECONDS = 1.0
MAX_CONSECUTIVE_CHUNK_FAILURES = 2

REDOWNLOAD_COMPLETED_CHUNKS = False
OVERWRITE_STANDARDISED_OUTPUTS = True

# Save only complete site CSVs.
EXPORT_PARTIAL_SITES = False

# False for notebooks; True for strict runs.
STRICT_COMPLETION = False

# Skip bad time files.
MIN_REQUIRED_FIELD_COVERAGE = 0.50

if ACQUISITION_MODE not in {"api", "manual"}:
    raise ValueError("ACQUISITION_MODE must be 'api' or 'manual'.")

if EXTRACTION_START >= EXTRACTION_END:
    raise ValueError("EXTRACTION_START must be earlier than EXTRACTION_END.")

print("Acquisition mode:", ACQUISITION_MODE)
print("Extraction start:", EXTRACTION_START)
print("Extraction end (exclusive):", EXTRACTION_END)
print("Chunk sizes:", SITE_CHUNK_DAYS)
````

## 3. Field Helpers

````python
ARRIVAL_ALIASES = [
    "connectionTime",
    "connection_time",
    "connectiontime",
    "connectTime",
    "arrival_time",
    "arrivalTime",
]

DEPARTURE_ALIASES = [
    "disconnectTime",
    "disconnect_time",
    "disconnecttime",
    "departure_time",
    "departureTime",
]

DONE_CHARGING_ALIASES = [
    "doneChargingTime",
    "done_charging_time",
    "donechargingtime",
    "chargingEndTime",
]

ENERGY_ALIASES = [
    "kWhDelivered",
    "kwhDelivered",
    "kwh_delivered",
    "energy_delivered_kwh",
    "energyDelivered",
    "energy_kwh",
]


def canonical_name(value: str) -> str:
    return re.sub(r"[^a-z0-9]", "", str(value).lower())


def resolve_column(
    frame: pd.DataFrame,
    aliases: list[str],
    nested_suffixes: tuple[str, ...] = ("date", "value", "timestamp"),
) -> str | None:
    columns = list(frame.columns)

    for alias in aliases:
        if alias in columns:
            return alias

    lower_map = {str(column).lower(): column for column in columns}

    for alias in aliases:
        match = lower_map.get(alias.lower())
        if match is not None:
            return match

    canonical_map: dict[str, list[str]] = {}

    for column in columns:
        canonical_map.setdefault(canonical_name(column), []).append(column)

    for alias in aliases:
        base = canonical_name(alias)

        for key in [base, *[base + suffix for suffix in nested_suffixes]]:
            matches = canonical_map.get(key, [])
            if matches:
                return matches[0]

    return None


def unwrap_datetime(value: Any) -> Any:
    if value is None:
        return None

    if isinstance(value, (pd.Timestamp, datetime, np.datetime64)):
        return value

    if isinstance(value, np.generic):
        value = value.item()

    if isinstance(value, dict):
        for key in (
            "$date",
            "date",
            "datetime",
            "timestamp",
            "value",
            "$numberLong",
        ):
            if key in value:
                return unwrap_datetime(value[key])

        if len(value) == 1:
            return unwrap_datetime(next(iter(value.values())))

        return None

    if isinstance(value, (list, tuple)):
        return unwrap_datetime(value[0]) if len(value) == 1 else None

    if isinstance(value, str):
        text = value.strip()

        if not text or text.lower() in {
            "nan",
            "nat",
            "none",
            "null",
            "<na>",
        }:
            return None

        if text[0] in "[{(" and text[-1] in "]})":
            for loader in (json.loads, python_ast.literal_eval):
                try:
                    return unwrap_datetime(loader(text))
                except Exception:
                    continue

        return text

    return value


def parse_datetime_scalar(value: Any) -> pd.Timestamp:
    value = unwrap_datetime(value)

    if value is None:
        return pd.NaT

    if isinstance(value, (int, float, np.integer, np.floating)):
        if pd.isna(value):
            return pd.NaT

        absolute = abs(float(value))

        try:
            if absolute >= 1e14:
                return pd.Timestamp(pd.to_datetime(value, unit="us", utc=True))
            if absolute >= 1e11:
                return pd.Timestamp(pd.to_datetime(value, unit="ms", utc=True))
            if absolute >= 1e8:
                return pd.Timestamp(pd.to_datetime(value, unit="s", utc=True))
            return pd.NaT
        except Exception:
            return pd.NaT

    if isinstance(value, str):
        try:
            parsed_email = parsedate_to_datetime(value)
            if parsed_email is not None:
                if parsed_email.tzinfo is None:
                    parsed_email = parsed_email.replace(tzinfo=timezone.utc)
                return pd.Timestamp(parsed_email).tz_convert("UTC")
        except (TypeError, ValueError, IndexError):
            pass

    try:
        parsed = pd.to_datetime(value, errors="coerce", utc=True)
    except Exception:
        return pd.NaT

    return pd.NaT if pd.isna(parsed) else pd.Timestamp(parsed)


def parse_datetime_series(values: pd.Series) -> pd.Series:
    parsed = values.apply(parse_datetime_scalar)
    return pd.to_datetime(parsed, errors="coerce", utc=True)


def first_available_series(
    frame: pd.DataFrame,
    aliases: list[str],
    default: Any = np.nan,
) -> pd.Series:
    column = resolve_column(frame, aliases)

    if column is None:
        return pd.Series(default, index=frame.index)

    return frame[column]


def json_safe(value: Any) -> Any:
    if isinstance(value, (datetime, pd.Timestamp)):
        return value.isoformat()
    if isinstance(value, np.generic):
        return value.item()
    if isinstance(value, Path):
        return str(value)
    return str(value)


def normalise_user_inputs(value: Any) -> str:
    if value is None or (
        isinstance(value, float) and np.isnan(value)
    ):
        return "[]"

    if isinstance(value, str):
        text = value.strip()

        if not text or text.lower() in {"nan", "none", "null", "<na>"}:
            return "[]"

        for loader in (json.loads, python_ast.literal_eval):
            try:
                value = loader(text)
                break
            except Exception:
                continue

    if isinstance(value, dict):
        value = [value]

    if not isinstance(value, list):
        return "[]"

    cleaned = [item for item in value if isinstance(item, dict)]
    return json.dumps(cleaned, default=json_safe, ensure_ascii=False)
````

## 4. API Setup

````python
class ACNAPIError(RuntimeError):
    pass


class ACNAuthenticationError(ACNAPIError):
    pass


class ACNResponseError(ACNAPIError):
    pass


REQUEST_AUDIT_ROWS: list[dict[str, Any]] = []
http_session: requests.Session | None = None

if ACQUISITION_MODE == "api":
    api_token = os.getenv("ACN_API_TOKEN", "").strip()

    if not api_token:
        api_token = getpass("Enter your ACN-Data API token: ").strip()

    if not api_token:
        raise EnvironmentError(
            "No ACN API token was supplied. "
            "Set ACN_API_TOKEN or enter the token in the hidden prompt."
        )

    http_session = requests.Session()
    http_session.auth = (api_token, "")
    http_session.headers.update(
        {
            "Accept": "application/json",
            "User-Agent": "SEND-EV-Project/Notebook-11A",
        }
    )

    print("Authenticated API session created.")
else:
    print("API session is not required in manual mode.")


def utc_rfc1123(value: datetime) -> str:
    if value.tzinfo is None:
        value = value.replace(tzinfo=timezone.utc)

    return format_datetime(value.astimezone(timezone.utc), usegmt=True)


def build_where_condition(
    start: datetime,
    end: datetime,
    min_energy: float,
) -> str:
    return (
        f'connectionTime>="{utc_rfc1123(start)}" '
        f'and connectionTime<"{utc_rfc1123(end)}" '
        f"and kWhDelivered>={min_energy}"
    )


def retry_delay(
    attempt: int,
    retry_after: str | None = None,
) -> float:
    if retry_after:
        try:
            return min(MAX_RETRY_SECONDS, max(0.0, float(retry_after)))
        except ValueError:
            pass

    return min(
        MAX_RETRY_SECONDS,
        BASE_RETRY_SECONDS * (2 ** (attempt - 1)) + random.uniform(0, 2),
    )


def response_preview(
    response: requests.Response,
    limit: int = 300,
) -> str:
    return (
        (response.text or "")
        .replace("\n", " ")
        .replace("\r", " ")[:limit]
    )


def request_json(
    session: requests.Session,
    url: str,
    params: dict[str, Any] | None,
    context: str,
) -> dict[str, Any]:
    last_error: Exception | None = None

    for attempt in range(1, MAX_REQUEST_ATTEMPTS + 1):
        response: requests.Response | None = None
        started = time.perf_counter()

        try:
            response = session.get(
                url,
                params=params,
                timeout=(CONNECT_TIMEOUT_SECONDS, READ_TIMEOUT_SECONDS),
            )

            elapsed = time.perf_counter() - started
            status = response.status_code
            content_type = response.headers.get("Content-Type", "")
            preview = response_preview(response)

            audit = {
                "timestamp_utc": datetime.now(timezone.utc).isoformat(),
                "context": context,
                "attempt": attempt,
                "requested_url": response.url,
                "status_code": status,
                "content_type": content_type,
                "response_bytes": len(response.content),
                "elapsed_seconds": elapsed,
                "response_preview": preview,
            }

            if status in {401, 403}:
                audit["outcome"] = "authentication_failed"
                REQUEST_AUDIT_ROWS.append(audit)
                raise ACNAuthenticationError(
                    f"Authentication failed with HTTP {status}."
                )

            if status == 429 or 500 <= status <= 599:
                audit["outcome"] = "retryable_http_error"
                REQUEST_AUDIT_ROWS.append(audit)
                raise ACNResponseError(f"Retryable HTTP status {status}.")

            response.raise_for_status()

            if not response.content:
                audit["outcome"] = "empty_response"
                REQUEST_AUDIT_ROWS.append(audit)
                raise ACNResponseError("The API returned an empty body.")

            try:
                payload = response.json()
            except requests.exceptions.JSONDecodeError as exc:
                audit["outcome"] = "invalid_json"
                REQUEST_AUDIT_ROWS.append(audit)
                raise ACNResponseError(
                    "The API returned non-JSON content. "
                    f"Content-Type={content_type!r}; preview={preview!r}"
                ) from exc

            if not isinstance(payload, dict):
                audit["outcome"] = "unexpected_json_type"
                REQUEST_AUDIT_ROWS.append(audit)
                raise ACNResponseError(
                    f"Expected a JSON object, received {type(payload).__name__}."
                )

            audit["outcome"] = "success"
            REQUEST_AUDIT_ROWS.append(audit)
            return payload

        except ACNAuthenticationError:
            raise

        except (requests.exceptions.RequestException, ACNResponseError) as exc:
            last_error = exc

            if response is None:
                REQUEST_AUDIT_ROWS.append(
                    {
                        "timestamp_utc": datetime.now(timezone.utc).isoformat(),
                        "context": context,
                        "attempt": attempt,
                        "requested_url": url,
                        "status_code": np.nan,
                        "content_type": None,
                        "response_bytes": np.nan,
                        "elapsed_seconds": time.perf_counter() - started,
                        "response_preview": None,
                        "outcome": f"request_exception:{type(exc).__name__}",
                    }
                )

            if attempt == MAX_REQUEST_ATTEMPTS:
                break

            delay = retry_delay(
                attempt,
                (
                    response.headers.get("Retry-After")
                    if response is not None
                    else None
                ),
            )

            print(
                f"{context}: attempt {attempt}/{MAX_REQUEST_ATTEMPTS} "
                f"failed ({type(exc).__name__}: {exc}). "
                f"Retrying in {delay:.1f}s."
            )
            time.sleep(delay)

    raise ACNAPIError(
        f"{context} failed after {MAX_REQUEST_ATTEMPTS} attempts. "
        f"Last error: {type(last_error).__name__}: {last_error}"
    ) from last_error
````

## 5. API Download

````python
def generate_time_chunks(
    start: datetime,
    end: datetime,
    chunk_days: int,
) -> Iterator[tuple[datetime, datetime]]:
    if start >= end:
        raise ValueError("Chunk start must be earlier than chunk end.")

    current = start

    while current < end:
        chunk_end = min(current + timedelta(days=chunk_days), end)
        yield current, chunk_end
        current = chunk_end


def chunk_name(
    site: str,
    start: datetime,
    end: datetime,
) -> str:
    return (
        f"{site}_{start:%Y%m%dT%H%M%S}_"
        f"{end:%Y%m%dT%H%M%S}.jsonl"
    )


def count_jsonl(path: Path) -> int:
    if not path.exists():
        return 0

    with path.open("r", encoding="utf-8") as handle:
        return sum(1 for line in handle if line.strip())


def sha256_file(
    path: Path,
    block_size: int = 1024 * 1024,
) -> str:
    digest = hashlib.sha256()

    with path.open("rb") as handle:
        while block := handle.read(block_size):
            digest.update(block)

    return digest.hexdigest()


def atomic_json(payload: dict[str, Any], path: Path) -> None:
    temporary = path.with_suffix(path.suffix + ".tmp")

    with temporary.open("w", encoding="utf-8") as handle:
        json.dump(payload, handle, indent=2, default=json_safe)

    temporary.replace(path)


def download_chunk(
    session: requests.Session,
    site: str,
    start: datetime,
    end: datetime,
    output_path: Path,
) -> dict[str, Any]:
    output_path.parent.mkdir(parents=True, exist_ok=True)

    part_path = output_path.with_suffix(output_path.suffix + ".part")
    checkpoint_path = output_path.with_suffix(
        output_path.suffix + ".checkpoint.json"
    )

    if REDOWNLOAD_COMPLETED_CHUNKS:
        output_path.unlink(missing_ok=True)
        part_path.unlink(missing_ok=True)
        checkpoint_path.unlink(missing_ok=True)

    if output_path.exists():
        return {
            "site": site,
            "chunk_start_utc": start,
            "chunk_end_utc": end,
            "chunk_file": output_path.name,
            "chunk_sessions": count_jsonl(output_path),
            "chunk_status": "reused_complete",
            "pages_completed": np.nan,
            "error": None,
        }

    endpoint = urljoin(API_BASE_URL, f"sessions/{site}")
    initial_params = {
        "where": build_where_condition(start, end, MIN_ENERGY_KWH),
        "sort": SORT_ARGUMENT,
    }

    if checkpoint_path.exists() and part_path.exists():
        checkpoint = json.loads(
            checkpoint_path.read_text(encoding="utf-8")
        )
        current_url = checkpoint["next_url"]
        current_params = None
        pages_completed = int(checkpoint["pages_completed"])
        downloaded = int(checkpoint["downloaded_sessions"])
        file_mode = "a"

        print(
            f"{site}: resuming {output_path.name} "
            f"after {pages_completed} pages."
        )
    else:
        part_path.unlink(missing_ok=True)
        checkpoint_path.unlink(missing_ok=True)
        current_url = endpoint
        current_params = initial_params
        pages_completed = 0
        downloaded = 0
        file_mode = "w"

    followed_urls: set[str] = set()

    with part_path.open(file_mode, encoding="utf-8") as output_file:
        while True:
            page_number = pages_completed + 1

            payload = request_json(
                session=session,
                url=current_url,
                params=current_params,
                context=(
                    f"{site}:{start.date()}:{end.date()}:page{page_number}"
                ),
            )

            items = payload.get("_items")

            if not isinstance(items, list):
                raise ACNResponseError(
                    "The response does not contain an `_items` list."
                )

            for record in items:
                output_file.write(
                    json.dumps(
                        record,
                        ensure_ascii=False,
                        default=json_safe,
                    )
                    + "\n"
                )
                downloaded += 1

            output_file.flush()
            os.fsync(output_file.fileno())
            pages_completed += 1

            next_href = (
                payload.get("_links", {})
                .get("next", {})
                .get("href")
            )

            if not next_href:
                break

            next_url = urljoin(API_BASE_URL, next_href)

            if next_url in followed_urls:
                raise ACNResponseError(
                    f"Pagination cycle detected at {next_url}."
                )

            followed_urls.add(next_url)

            atomic_json(
                {
                    "site": site,
                    "chunk_start_utc": start,
                    "chunk_end_utc": end,
                    "next_url": next_url,
                    "pages_completed": pages_completed,
                    "downloaded_sessions": downloaded,
                    "updated_at_utc": datetime.now(timezone.utc),
                },
                checkpoint_path,
            )

            current_url = next_url
            current_params = None
            time.sleep(PAUSE_BETWEEN_PAGES_SECONDS)

    part_path.replace(output_path)
    checkpoint_path.unlink(missing_ok=True)

    return {
        "site": site,
        "chunk_start_utc": start,
        "chunk_end_utc": end,
        "chunk_file": output_path.name,
        "chunk_sessions": downloaded,
        "chunk_status": "downloaded",
        "pages_completed": pages_completed,
        "error": None,
    }


def merge_complete_chunks(
    chunk_paths: list[Path],
    output_path: Path,
) -> tuple[int, int]:
    temporary = output_path.with_suffix(output_path.suffix + ".part")
    seen: set[str] = set()
    written = 0
    duplicates = 0

    with temporary.open("w", encoding="utf-8") as output_file:
        for chunk_path in sorted(chunk_paths):
            with chunk_path.open("r", encoding="utf-8") as input_file:
                for line in input_file:
                    if not line.strip():
                        continue

                    record = json.loads(line)
                    identifier = (
                        record.get("sessionID")
                        or record.get("_id")
                        or record.get("id")
                        or hashlib.sha256(
                            line.encode("utf-8")
                        ).hexdigest()
                    )
                    identifier = str(identifier)

                    if identifier in seen:
                        duplicates += 1
                        continue

                    seen.add(identifier)
                    output_file.write(line)
                    written += 1

    temporary.replace(output_path)
    return written, duplicates


def acquire_site_from_api(
    site: str,
    session: requests.Session,
) -> tuple[dict[str, Any], pd.DataFrame, Path | None]:
    windows = list(
        generate_time_chunks(
            EXTRACTION_START,
            EXTRACTION_END,
            SITE_CHUNK_DAYS[site],
        )
    )

    expected_paths = [
        CHUNK_DIR
        / site
        / chunk_name(site, chunk_start, chunk_end)
        for chunk_start, chunk_end in windows
    ]

    audit_rows: list[dict[str, Any]] = []
    consecutive_failures = 0

    for (chunk_start, chunk_end), chunk_path in zip(
        windows,
        expected_paths,
    ):
        try:
            result = download_chunk(
                session=session,
                site=site,
                start=chunk_start,
                end=chunk_end,
                output_path=chunk_path,
            )
            audit_rows.append(result)
            consecutive_failures = 0
            time.sleep(PAUSE_BETWEEN_CHUNKS_SECONDS)

        except ACNAuthenticationError as exc:
            audit_rows.append(
                {
                    "site": site,
                    "chunk_start_utc": chunk_start,
                    "chunk_end_utc": chunk_end,
                    "chunk_file": chunk_path.name,
                    "chunk_sessions": 0,
                    "chunk_status": "authentication_failed",
                    "pages_completed": np.nan,
                    "error": f"{type(exc).__name__}: {exc}",
                }
            )
            break

        except Exception as exc:
            audit_rows.append(
                {
                    "site": site,
                    "chunk_start_utc": chunk_start,
                    "chunk_end_utc": chunk_end,
                    "chunk_file": chunk_path.name,
                    "chunk_sessions": 0,
                    "chunk_status": "failed",
                    "pages_completed": np.nan,
                    "error": f"{type(exc).__name__}: {exc}",
                }
            )
            consecutive_failures += 1

            print(
                f"{site}: failed {chunk_start.date()} to "
                f"{chunk_end.date()}: {type(exc).__name__}: {exc}"
            )

            if consecutive_failures >= MAX_CONSECUTIVE_CHUNK_FAILURES:
                print(
                    f"{site}: stopping after {consecutive_failures} "
                    "consecutive failures. Completed chunks are preserved."
                )
                break

    completed_paths = [
        path for path in expected_paths if path.exists()
    ]
    complete = len(completed_paths) == len(expected_paths)
    raw_output = ACQUIRED_DIR / f"{site}_2019_raw.jsonl"

    written = 0
    duplicates = 0

    if complete or (EXPORT_PARTIAL_SITES and completed_paths):
        written, duplicates = merge_complete_chunks(
            completed_paths,
            raw_output,
        )
    else:
        raw_output = None

    summary = {
        "site": site,
        "mode": "api",
        "site_complete": complete,
        "expected_chunks": len(expected_paths),
        "completed_chunks": len(completed_paths),
        "raw_sessions": written,
        "duplicates_removed": duplicates,
        "raw_file": raw_output.name if raw_output else None,
        "error": (
            None
            if complete
            else "The configured annual extraction is incomplete."
        ),
    }

    return summary, pd.DataFrame(audit_rows), raw_output
````

## 6. Manual Files

````python
def load_source_file(path: Path) -> pd.DataFrame:
    suffix = path.suffix.lower()

    if suffix == ".csv":
        return pd.read_csv(path)

    if suffix == ".json":
        payload = json.loads(path.read_text(encoding="utf-8"))

        if isinstance(payload, list):
            records = payload
        elif isinstance(payload, dict) and isinstance(
            payload.get("_items"),
            list,
        ):
            records = payload["_items"]
        elif isinstance(payload, dict):
            records = [payload]
        else:
            raise ValueError(
                f"Unsupported JSON structure: {type(payload).__name__}"
            )

        return pd.json_normalize(records, sep=".")

    if suffix == ".jsonl":
        records: list[dict[str, Any]] = []

        with path.open("r", encoding="utf-8") as handle:
            for line_number, line in enumerate(handle, start=1):
                if not line.strip():
                    continue

                payload = json.loads(line)

                if isinstance(payload, dict) and isinstance(
                    payload.get("_items"),
                    list,
                ):
                    records.extend(payload["_items"])
                elif isinstance(payload, dict):
                    records.append(payload)
                else:
                    raise ValueError(
                        f"Unsupported JSONL record at line {line_number}."
                    )

        return pd.json_normalize(records, sep=".")

    raise ValueError(f"Unsupported file format: {path.suffix}")


def manual_candidates(site: str) -> list[Path]:
    return [
        MANUAL_DIR / f"{site}_2019_sessions.csv",
        MANUAL_DIR / f"{site}_2019_raw.json",
        MANUAL_DIR / f"{site}_2019_raw.jsonl",
        ACN_DIR / f"{site}_2019_sessions.csv",
        ACN_DIR / f"{site}_2019_raw.json",
        ACN_DIR / f"{site}_2019_raw.jsonl",
    ]


def stable_session_id(
    site: str,
    station: Any,
    arrival: Any,
    departure: Any,
    energy: Any,
) -> str:
    source = "|".join(
        [
            str(site),
            str(station),
            str(arrival),
            str(departure),
            str(energy),
        ]
    )
    return "derived-" + hashlib.sha256(
        source.encode("utf-8")
    ).hexdigest()[:24]


def standardise_sessions(
    source: pd.DataFrame,
    site: str,
    source_name: str,
) -> tuple[pd.DataFrame, dict[str, Any]]:
    source = source.copy()
    source.columns = [
        str(column).replace("\ufeff", "").strip()
        for column in source.columns
    ]

    arrival_column = resolve_column(source, ARRIVAL_ALIASES)
    departure_column = resolve_column(source, DEPARTURE_ALIASES)
    done_column = resolve_column(source, DONE_CHARGING_ALIASES)
    energy_column = resolve_column(source, ENERGY_ALIASES)

    if arrival_column is None:
        raise ValueError("No connection/arrival timestamp column was found.")
    if departure_column is None:
        raise ValueError("No disconnection/departure timestamp column was found.")
    if energy_column is None:
        raise ValueError("No delivered-energy column was found.")

    arrival = parse_datetime_series(source[arrival_column])
    departure = parse_datetime_series(source[departure_column])
    done = (
        parse_datetime_series(source[done_column])
        if done_column is not None
        else pd.Series(pd.NaT, index=source.index, dtype="datetime64[ns, UTC]")
    )
    energy = pd.to_numeric(source[energy_column], errors="coerce")

    arrival_share = float(arrival.notna().mean()) if len(source) else 0.0
    departure_share = float(departure.notna().mean()) if len(source) else 0.0
    energy_share = float(energy.notna().mean()) if len(source) else 0.0

    if min(arrival_share, departure_share, energy_share) < MIN_REQUIRED_FIELD_COVERAGE:
        raise ValueError(
            "Required-field coverage is too low: "
            f"arrival={arrival_share:.3f}, "
            f"departure={departure_share:.3f}, "
            f"energy={energy_share:.3f}."
        )

    session_id = first_available_series(
        source,
        ["sessionID", "session_id", "sessionId", "_id", "id"],
        default=None,
    ).astype("string")

    station_id = first_available_series(
        source,
        ["stationID", "station_id", "stationId"],
        default=None,
    ).astype("string")

    output = pd.DataFrame(
        {
            "session_id": session_id,
            "record_id": first_available_series(
                source,
                ["_id", "id", "record_id"],
                default=None,
            ).astype("string"),
            "site": site,
            "station_id": station_id,
            "space_id": first_available_series(
                source,
                ["spaceID", "space_id", "spaceId"],
                default=None,
            ).astype("string"),
            "cluster_id": first_available_series(
                source,
                ["clusterID", "cluster_id", "clusterId"],
                default=None,
            ).astype("string"),
            "user_id": first_available_series(
                source,
                ["userID", "user_id", "userId"],
                default=None,
            ).astype("string"),
            "connection_time_utc": arrival,
            "disconnect_time_utc": departure,
            "done_charging_time_utc": done,
            "kwh_delivered": energy,
            "source_max_power_kw": pd.to_numeric(
                first_available_series(
                    source,
                    [
                        "maximum_power_kw",
                        "max_power_kw",
                        "charger_power_kw",
                        "pilot_max_kw",
                    ],
                ),
                errors="coerce",
            ),
            "user_inputs_json": first_available_series(
                source,
                ["userInputs", "user_inputs"],
                default=None,
            ).apply(normalise_user_inputs),
            "source_file": source_name,
            "source_row_number": np.arange(len(source)),
            "extraction_start_utc": EXTRACTION_START.isoformat(),
            "extraction_end_utc_exclusive": EXTRACTION_END.isoformat(),
        }
    )

    missing_id = (
        output["session_id"].isna()
        | output["session_id"].str.strip().isin(
            ["", "<NA>", "nan", "None"]
        )
    )

    output.loc[missing_id, "session_id"] = [
        stable_session_id(
            site=site,
            station=station,
            arrival=arrival_value,
            departure=departure_value,
            energy=energy_value,
        )
        for station, arrival_value, departure_value, energy_value in zip(
            output.loc[missing_id, "station_id"],
            output.loc[missing_id, "connection_time_utc"],
            output.loc[missing_id, "disconnect_time_utc"],
            output.loc[missing_id, "kwh_delivered"],
        )
    ]

    output["session_id_derived"] = missing_id

    # Keep UTC times stable.
    for column in [
        "connection_time_utc",
        "disconnect_time_utc",
        "done_charging_time_utc",
    ]:
        output[column] = output[column].apply(
            lambda value: (
                value.isoformat()
                if pd.notna(value)
                else None
            )
        )

    diagnostics = {
        "source_rows": len(source),
        "standardised_rows": len(output),
        "arrival_parse_share": arrival_share,
        "departure_parse_share": departure_share,
        "energy_numeric_share": energy_share,
        "arrival_source_column": arrival_column,
        "departure_source_column": departure_column,
        "energy_source_column": energy_column,
        "derived_session_ids": int(missing_id.sum()),
    }

    return output, diagnostics


def acquire_site_from_manual(
    site: str,
) -> tuple[dict[str, Any], pd.DataFrame, Path | None]:
    audit_rows: list[dict[str, Any]] = []

    for candidate in manual_candidates(site):
        if not candidate.exists():
            audit_rows.append(
                {
                    "site": site,
                    "candidate": str(candidate.relative_to(ROOT)),
                    "exists": False,
                    "selected": False,
                    "status": "missing",
                    "rows": 0,
                    "error": None,
                }
            )
            continue

        try:
            source = load_source_file(candidate)

            if source.empty:
                raise ValueError("The file contains zero rows.")

            standardised, diagnostics = standardise_sessions(
                source=source,
                site=site,
                source_name=candidate.name,
            )

            output_path = ACQUIRED_DIR / f"{site}_2019_sessions.csv"
            standardised.to_csv(output_path, index=False)

            audit_rows.append(
                {
                    "site": site,
                    "candidate": str(candidate.relative_to(ROOT)),
                    "exists": True,
                    "selected": True,
                    "status": "standardised",
                    "rows": len(source),
                    "error": None,
                    **diagnostics,
                }
            )

            summary = {
                "site": site,
                "mode": "manual",
                "site_complete": True,
                "expected_chunks": np.nan,
                "completed_chunks": np.nan,
                "raw_sessions": len(source),
                "duplicates_removed": 0,
                "raw_file": candidate.name,
                "standardised_file": output_path.name,
                "standardised_sessions": len(standardised),
                "error": None,
                **diagnostics,
            }

            return summary, pd.DataFrame(audit_rows), candidate

        except Exception as exc:
            audit_rows.append(
                {
                    "site": site,
                    "candidate": str(candidate.relative_to(ROOT)),
                    "exists": True,
                    "selected": False,
                    "status": "rejected",
                    "rows": 0,
                    "error": f"{type(exc).__name__}: {exc}",
                }
            )

    summary = {
        "site": site,
        "mode": "manual",
        "site_complete": False,
        "expected_chunks": np.nan,
        "completed_chunks": np.nan,
        "raw_sessions": 0,
        "duplicates_removed": 0,
        "raw_file": None,
        "standardised_file": None,
        "standardised_sessions": 0,
        "error": "No usable manual file was found.",
    }

    return summary, pd.DataFrame(audit_rows), None
````

## 7. Run

````python
acquisition_rows: list[dict[str, Any]] = []
audit_frames: list[pd.DataFrame] = []

for site in SITES:
    print()
    print("=" * 72)
    print(f"Acquiring {site}")
    print("=" * 72)

    if ACQUISITION_MODE == "api":
        assert http_session is not None

        summary, audit, raw_path = acquire_site_from_api(
            site=site,
            session=http_session,
        )

        if raw_path is not None:
            output_path = ACQUIRED_DIR / f"{site}_2019_sessions.csv"

            if output_path.exists() and not OVERWRITE_STANDARDISED_OUTPUTS:
                standardised = pd.read_csv(output_path)
                summary["standardised_file"] = output_path.name
                summary["standardised_sessions"] = len(standardised)
                summary["standardisation_action"] = "reused_existing"
            else:
                try:
                    raw_frame = load_source_file(raw_path)
                    standardised, diagnostics = standardise_sessions(
                        source=raw_frame,
                        site=site,
                        source_name=raw_path.name,
                    )
                    standardised.to_csv(output_path, index=False)

                    summary.update(
                        {
                            "standardised_file": output_path.name,
                            "standardised_sessions": len(standardised),
                            "standardisation_action": "created",
                            **diagnostics,
                        }
                    )

                except Exception as exc:
                    summary.update(
                        {
                            "site_complete": False,
                            "standardised_file": None,
                            "standardised_sessions": 0,
                            "standardisation_action": "failed",
                            "error": (
                                f"Standardisation failed: "
                                f"{type(exc).__name__}: {exc}"
                            ),
                        }
                    )
        else:
            summary["standardised_file"] = None
            summary["standardised_sessions"] = 0
            summary["standardisation_action"] = "not_created"

    else:
        summary, audit, _ = acquire_site_from_manual(site)

    acquisition_rows.append(summary)

    if not audit.empty:
        audit_frames.append(audit)

acquisition_manifest = pd.DataFrame(acquisition_rows)
display(acquisition_manifest)

acquisition_manifest_path = ACQUIRED_DIR / "acquisition_manifest.csv"
acquisition_manifest.to_csv(acquisition_manifest_path, index=False)
print("Manifest saved:", acquisition_manifest_path)

acquisition_audit = (
    pd.concat(audit_frames, ignore_index=True, sort=False)
    if audit_frames
    else pd.DataFrame()
)

if not acquisition_audit.empty:
    audit_path = ACQUIRED_DIR / "acquisition_chunk_audit.csv"
    acquisition_audit.to_csv(audit_path, index=False)
    print("Acquisition audit saved:", audit_path)

request_audit = pd.DataFrame(REQUEST_AUDIT_ROWS)

if not request_audit.empty:
    request_audit_path = ACQUIRED_DIR / "acquisition_request_audit.csv"
    request_audit.to_csv(request_audit_path, index=False)
    print("Request audit saved:", request_audit_path)
````

## 8. Check

````python
interface_rows: list[dict[str, Any]] = []

for site in SITES:
    path = ACQUIRED_DIR / f"{site}_2019_sessions.csv"

    if not path.exists():
        interface_rows.append(
            {
                "site": site,
                "file": path.name,
                "exists": False,
                "rows": 0,
                "required_columns_present": False,
                "arrival_parse_share": 0.0,
                "departure_parse_share": 0.0,
                "energy_numeric_share": 0.0,
                "ready_for_cleaning": False,
            }
        )
        continue

    frame = pd.read_csv(path)

    required = {
        "session_id",
        "site",
        "connection_time_utc",
        "disconnect_time_utc",
        "kwh_delivered",
    }

    required_present = required.issubset(frame.columns)

    if required_present and len(frame):
        arrival_share = float(
            pd.to_datetime(
                frame["connection_time_utc"],
                errors="coerce",
                utc=True,
            ).notna().mean()
        )
        departure_share = float(
            pd.to_datetime(
                frame["disconnect_time_utc"],
                errors="coerce",
                utc=True,
            ).notna().mean()
        )
        energy_share = float(
            pd.to_numeric(
                frame["kwh_delivered"],
                errors="coerce",
            ).notna().mean()
        )
    else:
        arrival_share = 0.0
        departure_share = 0.0
        energy_share = 0.0

    ready = bool(
        len(frame) > 0
        and required_present
        and min(
            arrival_share,
            departure_share,
            energy_share,
        ) >= MIN_REQUIRED_FIELD_COVERAGE
    )

    interface_rows.append(
        {
            "site": site,
            "file": path.name,
            "exists": True,
            "rows": len(frame),
            "required_columns_present": required_present,
            "arrival_parse_share": arrival_share,
            "departure_parse_share": departure_share,
            "energy_numeric_share": energy_share,
            "ready_for_cleaning": ready,
        }
    )

acquisition_interface = pd.DataFrame(interface_rows)
display(acquisition_interface)

ready_sites = acquisition_interface.loc[
    acquisition_interface["ready_for_cleaning"],
    "site",
].tolist()

missing_sites = [
    site for site in SITES if site not in ready_sites
]

print("Ready :", ", ".join(ready_sites) or "None")

if missing_sites:
    print(
        "Not ready:",
        ", ".join(missing_sites),
    )
    print(
        "Completed chunks and request audits have been preserved. "
        "Rerun API mode later, or switch to manual mode after placing "
        "complete files in Data/ACN/manual."
    )

if STRICT_COMPLETION and missing_sites:
    raise RuntimeError(
        "Acquisition is incomplete for: "
        + ", ".join(missing_sites)
    )
````
